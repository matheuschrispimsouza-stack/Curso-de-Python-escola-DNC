#Estrutura de controle
Permite que um programa execute diferentes conjuntos de instruções dependendo de uma condição
Em python, issso é feito com os comandos if(SE), elif(SENÃOSE), else(SENÃO).


```python
x = 50

if x == 5:
  print("x é 5")
else:
  print("x não é 5")
```

```python
a = 8

resto = a % 2

if resto == 0:
  print(a, "é par")
else:
  print(a, "é ímpar")
```

```python
b = 11

if b > 10:
  print(b, "é maior que 10")
elif b > 3:
  print(b, "é maior que 3")
elif b < 10:
  print(b, "é menor que 10")
else:
  print(b, "é igual a 10")
```
