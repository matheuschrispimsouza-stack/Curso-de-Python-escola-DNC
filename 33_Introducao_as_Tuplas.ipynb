# Tuplas
São coleções ordenadas de elementos, como listas, mas têm uma característica
importante: são imutáveis, ou seja, não podem ser modificadas após a criação.
* As tuplas são criadas ultilizando parênteses () e podem conter diferentes
  tipos de dados.

```python
tupla_numeros = (1,3,5)
tupla_variada = (1, 'bruno', 3.14)
```

```python
tupla_variada[1]
```

```python
tupla_variada.append(7)
```

```python
tupla_variada[1] = 'abacaxi'
```

```python
cidade_lista = ['Sao paulo', 'Rio de Janeiro', 'Curitiba']
cidade_tupla = ('Sao paulo', 'Rio de Janeiro', 'Curitiba')

print(cidade_lista[1], cidade_tupla[1])
```

```python
cidade_lista[1] = 'Vitoria'
cidade_tupla[1] = 'Vitoria'
```

```python
print(cidade_lista)

print(cidade_tupla)
```
