#Área do retengulo

```python
largura = float(input("Digite a largura do retangulo: "))
altura =  float(input("Digite a altura do retangulo: "))

area = largura*altura
perimetro = 2*(largura+altura)

print("Área:", area)
print("Perimetro:", perimetro)
```
