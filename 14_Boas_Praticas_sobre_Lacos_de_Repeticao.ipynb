#Evite laços infinitos

```python
a = 1

for i in range(30):
  if i % 3 == 0:
    print(i)
    a = a + 1
  else:
    a = a + 2
print(f"A:, {a}")
```

```python
# Repetição enquanto o contador for maior que zero
contador = 8

while contador > 0:
  contador = contador - 1
  print(contador)
```
