```python
# IF = SE
# ELIF = ELSE + IF => SENÃO PERMITE VERIFICAR MÚLTIPLAS CONDIÇÕES EM SERIE


```

```python
condicao = 3 < 6
condicao2 = 5 > 1

if condicao:
  # executo o codigo
  print("3 < 6")
elif condicao2:
  print("5 > 1")
else:
  print("Nenhuma condição foi satisfeita")
```

```python
num = int(input("Digite um número: "))

if num > 0:
  print("O número é positivo")
elif num < 0:
  print("O número é negativo")
else:
  print("O número é zero")






```

```python
saldo = 7835049

print("Menu")
print("1 - Ver saldo")
print("2 - Ralizar depósito")
print("3 - Sair")

opcao = int(input("Digite uma opcão do menu"))

if opcao == 1:
  print(f"\nSeu saldo é {saldo}")
elif opcao == 2:
  deposito = int(input(" Qual o valor a ser depositado ?"))
  saldo = saldo + deposito
  print(f"\nDeposito de R${deposito}, Concluido")
  print(f"Seu saldo é {saldo}")
elif opcao == 3:
  print("\nObrigado por usar nossos serviços")
else:
  print("\nOpção inválida")

```
