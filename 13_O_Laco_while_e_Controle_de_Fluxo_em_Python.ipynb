##Controle de fluxo
# while = Enquanto

#Break => interrompe o laço imediatamente e continua a execução do código após o laço.

#Continue => Pula a iteração atual do laço e vai para a próxima.

```python
i = 1

while i >= 20:
  if i % 2 == 0:
    i = i + 1
    continue
  print(f"Numero impar: {i}")
  i = i + 1
```

```python
contador = 11

while contador > 0:
  if contador % 3 == 0:
    break
  print(contador)
  contador = contador - 1
```

```python
contador = 11

while contador > 0:
  if contador % 3 == 0:
    contador = contador - 1
    continue

  print(contador)
  contador = contador - 1
```

```python
i = 1

while i <= 20:
  if i % 2 == 0:
    i = i + 1
    continue
  print(f"Numero impar: {i}")
  i = i + 1
```

```python
soma = 0

numero = 1

while True:
  soma = soma + numero
  if soma > 100:
    break
  numero = numero + 1

print(f"Soma final: {soma}")
```
