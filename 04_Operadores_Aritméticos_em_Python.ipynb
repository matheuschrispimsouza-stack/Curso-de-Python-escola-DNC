# Exercício
​Escreva um programa que tenha duas variáveis largura e altura que representem as dimensões de um retângulo. Use operadores aritméticos para calcular e exibir a área e o perímetro do retângulo.




```python
largura = 5
altura = 3
area = largura * altura
perimetro = 2 * (largura + altura)
print(f"Área do retangulo = {area}")
print(f"Perímetro do retangulo = {perimetro}")
```

# Calculadora

```python
x = 10
y = 3

print(f"Soma = {x + y}")
print(f"Subtração = {x - y}")
print(f"Multiplicação = {x * y}")
print(f"Divisão = {x / y}")
print(f"Divisão inteira = {x // y}")
print(f"Resto da Divisão = {x % y}")
print(f"Potência = {x ** y}")
```
