#.append
Adiciona um elemento ao final da lista

#.insert(indice,elemento)
Insere um elemento em uma posição específica, deslocando os elementos seguintes.

#.remove
remove o primeiro elemento com o valor especificado

#.pop
remove e retorna o elemento em uma determinada posição (ou o último, se não especificado).

#.Clear
remove todos os elementos, deixando a lista vazia


```python
list_nums = [1,2,3,4]
print(list_nums)
```

```python
#adicionar itens na lista
list_nums.append(5)
print(list_nums)
```

```python
list_nums.append('seis')
print(list_nums)
```

```python
list_nums.insert(0,'zero')
print(list_nums)
```

```python
list_nums.insert(2,'terceiro_elemento')
print(list_nums)
```

```python
list_nums.remove(1)
print(list_nums)
```

```python
list_nums.insert(1,1)
list_nums.insert(4,1)
print(list_nums)
```

```python
list_nums.remove(1)
print(list_nums)
```

```python
list_nums.pop()
```

```python
list_nums.clear()
```

```python
#alteração de valores de listas
lista_2 = [10, 20, 30]
print(lista_2)

lista_2[1] = 50
print(lista_2)
```

```python
# lista concatenada
lista_3 = [3, 76, 90]
print( lista_2 + lista_3 )
```

```python
numeros = [10, 20, 30, 40, 50]

numeros.append(60)
numeros.insert(1, 12)
numeros.remove(30)

print(numeros)
```

```python
itens = ['mesa', 'cadeira', 'armario']

itens[1] = 'poltrona'

removido = itens.pop()

itens.clear()

print(removido)
print(itens)
```
