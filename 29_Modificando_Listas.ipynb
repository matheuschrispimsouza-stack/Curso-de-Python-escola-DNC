#.append
Adiciona um elemento ao final da lista

#.insert(indice,elemento)
Insere um elemento em uma posição específica, deslocando os elementos seguintes.

#.remove
remove o primeiro elemento com o valor especificado

#.pop
remove e retorna o elemento em uma determinada posição (ou o último, se não especificado).


```python
list_nums = [1,2,3,4]
print(list_nums)
```

```python
#adicionar itens na lista
list_nums.append(5)
print(list_nums)
```

```python
list_nums.append('seis')
print(list_nums)
```

```python
list_nums.insert(0,'zero')
print(list_nums)
```

```python
list_nums.insert(2,'terceiro_elemento')
print(list_nums)
```

```python
list_nums.remove(1)
print(list_nums)
```

```python
list_nums.insert(1,1)
list_nums.insert(4,1)
print(list_nums)
```
