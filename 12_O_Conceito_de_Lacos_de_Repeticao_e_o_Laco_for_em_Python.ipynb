```python
soma = 0

for numero in range(1, 21):
  soma = soma + numero
  print(f"Soma parcial: {soma}")

print(soma)
```

```python
n = 5
fatorial = 1

for i in range(1 , n + 1):
  fatorial = fatorial * i
  print(f"Fatorial parcial: {fatorial}")

print(f"\nO fatorial de {5} é {fatorial}")
#
```
