```python
x = 7
y = 7


```

```python
eh_igual = x == y # Operador igual
eh_diferente = x != y # Operador diferente
eh_maior = x > y # Operador maior
eh_menor = x < y # Operador menor
eh_maior_igual = x >= y # Operador maior igual
eh_menor_igual = x <= y # Operador menor igual


print(eh_igual)
print(eh_diferente)
print(eh_maior)
print(eh_menor)
print(eh_maior_igual)
print(eh_menor_igual)
```

#Comparando idades

```python
idade_pessoa_1 = 26
idade_pessoa_2 = 26

sao_iguais = idade_pessoa_1 == idade_pessoa_2
sao_diferentes = idade_pessoa_1 != idade_pessoa_2

print(f"as idades são iguais? {sao_iguais}")
print(f"as idades são diferentes? {sao_diferentes}")
```

# Calculadora de IMC


```python
altura = 1.75
peso = 85

imc = peso / (altura ** 2)




print(f"IMC: {imc}")
print(f"Maior que 25: {imc > 25}")
print(f"Maior que 18,5: {imc < 18.5}")
```
